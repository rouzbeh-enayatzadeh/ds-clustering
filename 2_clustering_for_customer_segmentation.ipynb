# Clustering for Customer Segmentation

Customer Segmentation is the subdivision of a market into discrete customer groups that share similar characteristics. Customer Segmentation can be a powerful means to identify unsatisfied customer needs. Using the above data companies can then outperform the competition by developing uniquely appealing products and services.

The most common ways in which businesses segment their customer base are:
1. Demographic information, such as gender, age, familial and marital status, income, education, and occupation.
2. Geographical information, which differs depending on the scope of the company. For localized businesses, this info might pertain to specific towns or counties. For larger companies, it might mean a customer’s city, state, or even country of residence.
3. Psychographics, such as social class, lifestyle, and personality traits.
4. Behavioral data, such as spending and consumption habits, product/service usage, and desired benefits.

Advantages of Customer Segmentation
1. Determine appropriate product pricing.
2. Develop customized marketing campaigns.
3. Design an optimal distribution strategy.
4. Choose specific product features for deployment.
5. Prioritize new product development efforts.

## Clustering the Customer Data

We will cluster the data by Annual Income (k$) and Spending Score. So first we will create an array that only contains this two elements.

```python
x = df.iloc[:, [2, 3]].values
```

Now we will use the Elbow Method to get the best number of clusters. We saw in the plot above that the data already shows 5 clusters so we will use the Elbow Method in this case to validate the number of clusters we expect.

```python
wcss = []
for i in range(1, 11):
    km = KMeans(n_clusters=i, init="k-means++", max_iter=300, n_init=10, random_state=0)
    km.fit(x)
    wcss.append(km.inertia_)
fig = plt.figure(figsize=(10, 6))
fig.patch.set_facecolor("#f6f5f5")

plt.plot(range(1, 11), wcss)
plt.title("The Elbow Method", fontsize=20)
plt.xlabel("No. of Clusters")
plt.ylabel("wcss")
fig.text(0.5, 0.4, "The best k-value is 5")
plt.show()
```

The elbow method plots the within-cluster sum of squares (WCSS) against the number of clusters. WCSS always falls as k grows, so we look for the elbow: the point where the fall slows sharply. Here that bend is at k = 5, which matches the five groups visible in the income-versus-spending scatter plot. We therefore set `n_clusters = 5`.


Another way to choose k is to inspect the silhouette plots (sometimes called knife plots, because of their blade-like shape). Let's create silhouette plots for values of k from 2 to 10. The silhouette coefficient of a point measures how close it is to its own cluster compared to the nearest other cluster, on a scale from -1 to 1.


```python
# Range of k values
range_n_clusters = range(2, 11)

# Create a figure to hold the subplots
fig, axes = plt.subplots(5, 2, figsize=(15, 25))
axes = axes.flatten()

for k_idx, n_clusters in enumerate(range_n_clusters):
    ax = axes[k_idx]

    # Create a subplot for each k value
    ax.set_xlim([-0.1, 1])
    ax.set_ylim([0, len(x) + (n_clusters + 1) * 10])

    # Initialize the KMeans algorithm
    kmeans = KMeans(n_clusters=n_clusters, init="k-means++", max_iter=300, n_init=10, random_state=0)
    cluster_labels = kmeans.fit_predict(x)

    # Compute the silhouette scores for each sample
    silhouette_avg = silhouette_score(x, cluster_labels)
    sample_silhouette_values = silhouette_samples(x, cluster_labels)

    y_lower = 10
    for i in range(n_clusters):
        ith_cluster_silhouette_values = sample_silhouette_values[cluster_labels == i]
        ith_cluster_silhouette_values.sort()

        size_cluster_i = ith_cluster_silhouette_values.shape[0]
        y_upper = y_lower + size_cluster_i

        color = plt.cm.nipy_spectral(float(i) / n_clusters)
        ax.fill_betweenx(np.arange(y_lower, y_upper), 0, ith_cluster_silhouette_values, facecolor=color, edgecolor=color, alpha=0.7)

        ax.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))
        y_lower = y_upper + 10

    ax.set_title(f"Number of Clusters: {n_clusters}")
    ax.set_xlabel("Silhouette coefficient values")
    ax.set_ylabel("Cluster label")
    ax.axvline(x=silhouette_avg, color="red", linestyle="--")
    ax.set_yticks([])

# Adjust layout to avoid overlap
plt.tight_layout()
plt.show()
```

The average silhouette score (the red dashed line in each plot) is highest at k = 5, at about 0.55, and the five silhouettes are relatively even and mostly positive. This agrees with the elbow method, so k = 5 is a sound choice. A higher average silhouette means points sit more clearly inside their own cluster than in a neighbouring one.


```python
km = KMeans(n_clusters=5, init="k-means++", max_iter=300, n_init=10, random_state=0)
y_means = km.fit_predict(x)
```

For a quick overview how the clusters look like we can create a pandas series with the clusters and concatenate it onto the original dataset and print some samples for each cluster.

```python
df_cluster = pd.Series(y_means, name="Cluster")
```

```python
df_clustered = pd.concat([df, df_cluster], axis=1)
```

So of course our target group is the high income with high spending score. But we can now also think about how we get the high income with low spending score into the high spending group and conduct targeted marketing campaign. So who is in this group?

```python
# KMeans numbers the clusters arbitrarily, so we identify the high income, low spending
# cluster from the centroids instead of hard-coding an index.
centers = pd.DataFrame(km.cluster_centers_, columns=["Annual Income (k$)", "Spending Score (1-100)"])
high_income_low_spending_cluster = (centers["Annual Income (k$)"] - centers["Spending Score (1-100)"]).idxmax()

df_high_income_low_spending = df_clustered[["Age", "Gender", "Annual Income (k$)", "Spending Score (1-100)", "Cluster"]].query("Cluster == @high_income_low_spending_cluster")
```

```python
# new column based on clustering criterion to split to high income low spending group
df_clustered["income_group"] = df_clustered["Cluster"].apply(lambda x: 0 if x == high_income_low_spending_cluster else 1)
fig = plt.figure(figsize=(10, 6))
ax = sns.kdeplot(data=df_clustered, x="Age", hue="income_group", fill=True, legend=False, palette=["black", "#b20710"])

fig.patch.set_facecolor("whitesmoke")
ax.set_facecolor("white")

# ax.axes.get_xaxis().set_visible(False)
ax.axes.get_yaxis().set_visible(False)
ax.spines["bottom"].set_visible(False)
ax.spines["left"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["top"].set_visible(False)
fig.text(0.3, 1, "Distribution of Age", {"font": "Serif", "weight": "bold", "size": "22", "color": "black"}, alpha=0.7)
fig.text(0.2, 0.95, "Customers", **{"font": "serif", "size": 18, "weight": "bold", "color": "#b20710"}, alpha=0.6)
fig.text(0.2, 0.90, "Only the high income low spending score group", **{"font": "serif", "size": 18, "weight": "bold", "color": "black"}, alpha=0.6)

plt.show()
```

## Exercise

What happens when we include age and gender in the clustering? Do you still see clusters? If so, how are they composed? Feel free to visualise the results using different pairwise feature combinations, or even a 3D plot.


```python
from sklearn.preprocessing import StandardScaler

df_all = df.copy()
df_all["Gender_Encoded"] = df_all["Gender"].map({"Male": 0, "Female": 1})
df_all["Gender_Encoded"].value_counts()
```

```python
features = ["Age", "Gender_Encoded", "Annual Income (k$)", "Spending Score (1-100)"]
X_all = StandardScaler().fit_transform(df_all[features])

print(X_all.shape)
print(X_all.std(axis=0).round(2))
```

```python
for k in range(2, 11):
    labels = KMeans(n_clusters=k, init="k-means++", n_init=10, random_state=0).fit_predict(X_all)
    print(f"k={k}: silhouette={silhouette_score(X_all, labels):.3f}")
```

```python
labels_5 = KMeans(n_clusters=5, init="k-means++", n_init=10, random_state=0).fit_predict(X_all)
df_all["Cluster"] = labels_5
df_all.groupby("Cluster")[features].mean().round(1)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.scatterplot(data=df_all, x="Annual Income (k$)", y="Spending Score (1-100)", hue="Cluster", style="Gender", palette="viridis", ax=axes[0])
axes[0].set_title("Income vs Spending")
sns.scatterplot(data=df_all, x="Age", y="Spending Score (1-100)", hue="Cluster", style="Gender", palette="viridis", ax=axes[1])
axes[1].set_title("Age vs Spending")
sns.scatterplot(data=df_all, x="Age", y="Annual Income (k$)", hue="Cluster", style="Gender", palette="viridis", ax=axes[2])
axes[2].set_title("Age vs Income")
plt.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection="3d")
p = ax.scatter(df_all["Annual Income (k$)"], df_all["Spending Score (1-100)"], df_all["Age"], c=df_all["Cluster"], cmap="viridis", s=40, alpha=0.7)
ax.set_xlabel("Annual Income (k$)")
ax.set_ylabel("Spending Score")
ax.set_zlabel("Age")
fig.colorbar(p, label="Cluster", shrink=0.6)
plt.show()
```

### Conclusion of the exercise

**Do we still see clusters?** Yes, but they are much weaker. With all four scaled features the silhouette score at k=5 drops to 0.318 (vs 0.554 with income and spending only), and it keeps rising up to k=10 (0.421), so there is no clear best k.

**How are they composed?** Gender dominates. After scaling, the binary gender column puts every customer on one of two "floors" about 2 std apart, so K-Means splits the real groups into a male and a female copy: 4 of the 5 clusters are 100% one gender (clusters 1 vs 2 and 3 vs 4 differ almost only by gender). Only the high-income / low-spending group (cluster 0) stays intact and mixed.

| Features | Best k | Best silhouette |
|---|---|---|
| Income + Spending | 5 | 0.554 |
| Age + Income + Spending | 6 | 0.427 |
| Age + Gender + Income + Spending | none (rises to k=10) | 0.421 |

**Decision:** cluster on annual income and spending score only, and use age and gender afterwards to profile the segments (e.g. the high-spending customers are mostly younger than about 40).
